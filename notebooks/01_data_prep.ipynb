```python
import pandas as pd 
from sklearn.model_selection import train_test_split 

df = pd.read_csv("../data/Telco-Customer-Churn.csv") 

df 
```

## Dropping 'customerID' column since it carries No Predictive Signal 

```python
df = df.drop(columns = ['customerID']) 

df 
```

## Checking Data Types and Non-null counts 

```python
df.info() 
```

## Converting 'TotalCharges' column to numeric

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'] , errors = 'coerce')
```

```python
df.isnull().sum() 
```

```python
df['TotalCharges'] = df['TotalCharges'].fillna(0) 
```

```python
df.isnull().sum() 
```

## Map 'Yes'/'No' to 1/0 for Model Training 

```python
df['Churn'] = df['Churn'].map({'Yes' : 1 , 'No' : 0}) 

df['Churn'].value_counts(normalize = True) 
```

```python
x = df.drop(columns = ['Churn'])
y = df['Churn'] 
```

## Perform Stratified Train-Test Split to preserve Churn Ratio across Both Train and Test Sets

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 42 , stratify = y) 

print(f'Training Shape : {x_train.shape}')
print(f'Testing Shape : {x_test.shape}') 
```

## Identify Numerical and Categorical Columns 

```python
from sklearn.compose import ColumnTransformer 
from sklearn.preprocessing import OneHotEncoder , StandardScaler 
from sklearn.pipeline import Pipeline 

num_cols = ['tenure' , 'MonthlyCharges' , 'TotalCharges'] 
cat_cols = [col for col in x_train.columns if col not in num_cols] 

print('Numerical Features :' , num_cols) 
print('Categorical Features :' , cat_cols) 
```

## Preprocessing Pipeline 
### Define Transformers for Numerical and Categorical Data 

```python
num_transformer = Pipeline(steps = [
    ('scaler' , StandardScaler())
])

cat_transformer = Pipeline(steps = [
    ('onehot' , OneHotEncoder(drop = 'first' , sparse_output = False , handle_unknown = 'ignore') ) 
])

# Combine into a Single ColumnTransformer 

preprocessor = ColumnTransformer(
    transformers = [
        ('num' , num_transformer , num_cols),
        ('cat' , cat_transformer , cat_cols)
    ]
) 

# Fit Transform on Training Data and Only Transform on Testing Data 
x_train_processed = preprocessor.fit_transform(x_train) 
x_test_processed = preprocessor.transform(x_test) 

# Get encoded columns names to rebuild Dataframes for easy inspection 
encoded_cat_cols = preprocessor.named_transformers_['cat']['onehot'].get_feature_names_out(cat_cols) 
all_feature_names = num_cols + list(encoded_cat_cols) 

# Convert back to Pandas DataFrames 
x_train_prep = pd.DataFrame(x_train_processed , columns = all_feature_names) 
x_test_prep = pd.DataFrame(x_test_processed , columns = all_feature_names) 

x_train_prep 
```

## Save these Preprocessed Datasets 

```python
import os 

os.makedirs("../data/processed" , exist_ok = True) 

x_train_prep.to_csv("../data/processed/x_train.csv" , index = False)
x_test_prep.to_csv("../data/processed/x_test.csv" , index = False) 
y_train.to_csv("../data/processed/y_train.csv" , index = False) 
y_test.to_csv("../data/processed/y_test.csv" , index = False) 

print("Preprocessed Datasets Saved Successfully") 
```

```python
import joblib

joblib.dump(preprocessor, 'churn_preprocessor.joblib')
print("Preprocessor saved successfully!")
```
